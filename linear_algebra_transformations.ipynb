```python
# Make all numpy available via shorter 'np' prefix
import numpy as np
# Make all sympy available via shorter 'sym' prefix
import sympy as sym

import math

import matplotlib.pyplot as plt
%matplotlib inline

def plot2d(H, color='g', marker='ro', label=None):
    x = H[0,:]
    y = H[1,:]
    plt.plot(x, y, marker, x, y, color+'-', label=label)
    plt.axis([-10, 10, -10, 10])
    plt.grid()
```

# Geometry and Matrices

#### This question uses python to illustrate the geometric meaning of some special types of matrices. At next cell type:
- `plot2d(H)`

```python
H = np.array([[-6,-6,-7, 0, 7, 6, 6,-3,-3, 0, 0,-6],
              [-7, 2, 1, 8, 1, 2,-7,-7,-2,-2,-7,-7]])
print('H = \n', H)

plt.figure(figsize=(5,5))
plot2d(H)
plt.title('Original House')
plt.show()
```

## (a) Rotations

Generate a rotation matrix **Q** and let Q act on the house.

- `t = math.pi / 6`
- `Q = np.array([[math.cos(t), -math.sin(t)], [math.sin(t), math.cos(t)]])`
- Calculate `np.linalg.det(Q)`

Repeat with `t = -pi/3` and print the graph with **three** house images on the same figure.

```python
#Rotation 1:
t = math.pi / 6
Q = np.array([[math.cos(t), -math.sin(t)],
              [math.sin(t),  math.cos(t)]])
print('Q (t = pi/6) =\n', Q)
det_Q1 = np.linalg.det(Q)
print(f'det(Q) for t = pi/6 = {det_Q1:.6f}')
H_rot1 = Q @ H  

#Rotation 2:
t2 = -math.pi / 3
Q2 = np.array([[math.cos(t2), -math.sin(t2)],
               [math.sin(t2),  math.cos(t2)]])
print('\nQ2 (t = -pi/3) =\n', Q2)
det_Q2 = np.linalg.det(Q2)
print(f'det(Q2) for t = -pi/3 = {det_Q2:.6f}')
H_rot2 = Q2 @ H  

#Plot houses
plt.figure(figsize=(6, 6))

#Original house
plt.plot(H[0,:],      H[1,:],      'go', H[0,:],      H[1,:],      'g-',  label='Original')
#House rotated by pi/6
plt.plot(H_rot1[0,:], H_rot1[1,:], 'bs', H_rot1[0,:], H_rot1[1,:], 'b-',  label='Rotated t=π/6')
#House rotated by -pi/3
plt.plot(H_rot2[0,:], H_rot2[1,:], 'm^', H_rot2[0,:], H_rot2[1,:], 'm-',  label='Rotated t=-π/3')
plt.axis([-10, 10, -10, 10])
plt.grid(True)
plt.legend()
plt.title('(a) Rotations: Original and Two Rotated Houses')
plt.show()

print('det(Q) = 1 for t = pi/6  →', round(det_Q1, 6))
print('det(Q) = 1 for t = -pi/3 →', round(det_Q2, 6))
```

## (b) Dilations

Generate a dilation matrix **D** and let D act on the house.

- `r = .9`
- `D = np.array([[r, 0], [0, 1/r]])`
- Calculate `np.linalg.det(D)`

Repeat with `r = .8` and display the graph with **three** house images on the same figure.

```python
#Dilation 1: 
r = 0.9
D = np.array([[r, 0],
              [0, 1/r]])
print('D (r = 0.9) =\n', D)
det_D1 = np.linalg.det(D)
print(f'det(D) for r = 0.9 = {det_D1:.6f}')
H_dil1 = D @ H   

#Dilation 2
r2 = 0.8
D2 = np.array([[r2, 0],
               [0,  1/r2]])
print('\nD2 (r = 0.8) =\n', D2)
det_D2 = np.linalg.det(D2)
print(f'det(D2) for r = 0.8 = {det_D2:.6f}')
H_dil2 = D2 @ H 

#Plot houses
plt.figure(figsize=(6, 6))
# Original house
plt.plot(H[0,:],      H[1,:],      'go', H[0,:],      H[1,:],      'g-',  label='Original')
#House dilated by r=0.9
plt.plot(H_dil1[0,:], H_dil1[1,:], 'bs', H_dil1[0,:], H_dil1[1,:], 'b-',  label='Dilated r=0.9')
#House dilated by r=0.8
plt.plot(H_dil2[0,:], H_dil2[1,:], 'm^', H_dil2[0,:], H_dil2[1,:], 'm-',  label='Dilated r=0.8')
plt.axis([-10, 10, -10, 10])
plt.grid(True)
plt.legend()
plt.title('(b) Dilations: Original and Two Dilated Houses')
plt.show()

#Answers 
print('det(D) for r = 0.9 →', round(det_D1, 6))
print('det(D) for r = 0.8 →', round(det_D2, 6))
```

## (c) Shearing Transformations

Generate a shearing matrix **T** and let T act on the house.

- `t = 1/2`
- `T = np.array([[1, t], [0, 1]])`
- Calculate `np.linalg.det(T)`

Repeat with `t = -1/2`. Display **each** graph as the original house together with the transformed version.

```python
#Shear 1: 
t = 1/2
T = np.array([[1, t],
              [0, 1]])
print('T (t = 1/2) =\n', T)
det_T1 = np.linalg.det(T)
print(f'det(T) for t = 1/2 = {det_T1:.6f}')
H_shear1 = T @ H   

#Plot houses
plt.figure(figsize=(6, 6))
plt.plot(H[0,:],        H[1,:],        'go', H[0,:],        H[1,:],        'g-', label='Original')
plt.plot(H_shear1[0,:], H_shear1[1,:], 'bs', H_shear1[0,:], H_shear1[1,:], 'b-', label='Sheared t=1/2')
plt.axis([-10, 10, -10, 10])
plt.grid(True)
plt.legend()
plt.title('(c) Shearing: Original vs Sheared House (t = 1/2)')
plt.show()

#Shear 2:
t2 = -1/2
T2 = np.array([[1, t2],
               [0,  1]])
print('\nT2 (t = -1/2) =\n', T2)
det_T2 = np.linalg.det(T2)
print(f'det(T2) for t = -1/2 = {det_T2:.6f}')
H_shear2 = T2 @ H  

#Plot houses
plt.figure(figsize=(6, 6))
plt.plot(H[0,:],        H[1,:],        'go', H[0,:],        H[1,:],        'g-', label='Original')
plt.plot(H_shear2[0,:], H_shear2[1,:], 'm^', H_shear2[0,:], H_shear2[1,:], 'm-', label='Sheared t=-1/2')
plt.axis([-10, 10, -10, 10])
plt.grid(True)
plt.legend()
plt.title('(c) Shearing: Original vs Sheared House (t = -1/2)')
plt.show()

#Answers
print('det(T)  for t =  1/2 →', round(det_T1, 6))
print('det(T2) for t = -1/2 →', round(det_T2, 6))
```
